Identify where the reported coupling-coordination index improved or declined between 2010 and 2022.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
DATA_PATH = ROOT / 'data' / 'processed' / 'henan_city_system_indices_2010_2022.csv'
```

```python
from src.analysis import city_coupling_change, load_data

data = load_data(DATA_PATH)
changes = city_coupling_change(data)
changes
```

```python
pd.Series({
    'cities_improved': int((changes['change_2010_2022'] > 0).sum()),
    'cities_declined': int((changes['change_2010_2022'] < 0).sum()),
    'largest_gain_city': changes.iloc[0]['city'],
    'largest_gain': changes.iloc[0]['change_2010_2022'],
}).to_frame('result')
```
