# Temporal comparison

Compare provincial means while retaining the source report's original index scales.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
DATA_PATH = ROOT / 'data' / 'processed' / 'henan_city_system_indices_2010_2022.csv'
```

```python
from src.analysis import load_data, provincial_summary

data = load_data(DATA_PATH)
summary = provincial_summary(data)
summary
```

```python
start = summary.set_index('year').loc[2010]
end = summary.set_index('year').loc[2022]
pd.Series({
    'urbanisation_change_pct': 100 * (end['urbanisation_index_reported'] / start['urbanisation_index_reported'] - 1),
    'ecosystem_health_change_pct': 100 * (end['ecosystem_health_index_reported'] / start['ecosystem_health_index_reported'] - 1),
    'coupling_change_points': end['coupling_coordination_reported'] - start['coupling_coordination_reported'],
}).round(3)
```

Urbanisation rose strongly, ecosystem-health values were broadly flat to slightly lower, and mean coupling coordination improved overall but dipped slightly after 2020.
